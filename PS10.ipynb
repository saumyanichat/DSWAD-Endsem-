```python
import pandas as pd
```

```python
df=pd.read_csv("Iris.csv",header=None)
```

```python
print(df.head())
```

```python
print(df.head())
```

```python
df.columns = ['SepalLengthCm',
              'SepalWidthCm',
              'PetalLengthCm',
              'PetalWidthCm',
              'Species']
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
setosa_data = df[df['Species']=='Iris-setosa']
```

```python
versicolor_data = df[df['Species']=='Iris-versicolor']
```

```python
print(setosa_data.head())
```

```python
print(versicolor_data.head())
```

```python
df_concat = pd.concat([setosa_data,versicolor_data])
```

```python
print(df_concat)
```

```python
df_concat.head()
```

```python
df_sorted = df.sort_values(by='PetalLengthCm',ascending=True)
```

```python
df_sorted.head()
```

```python
df_tran = df.head().T
```

```python
print(df_tran)
```

```python
melted_data = pd.melt(df,
                      id_vars=['Species'],
                      value_vars=['SepalLengthCm',
                                  'SepalWidthCm',
                                  'PetalLengthCm',
                                  'PetalWidthCm'],
                      var_name='measurement',
                      value_name='value')
```

```python
print(melted_data.head())
```

```python
wide = melted_data.pivot_table(index='Species',
                               columns='measurement',
                               values='value',
                               aggfunc='mean')
```
