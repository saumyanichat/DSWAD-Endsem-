```python
melted_data=pd.melt(
    df,
    id_vars=['Type'],
    value_vars=['like','share','comment'],
    var_name='metric',value_name='count')
```

```python
print(melted_data.head())
```

```python
wide=melted_data.pivot_table(index='Type' ,columns='metric', values='count' ,aggfunc='mean')
```
