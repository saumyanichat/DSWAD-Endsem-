```python
melted_data = pd.melt(df,
                      id_vars=['native_country'],
                      value_vars=['age',
                                  'hours_per_week',
                                  'education_num'],
                      var_name='measurement',
                      value_name='value')
```

```python
print(melted_data.head())
```

```python
wide = melted_data.pivot_table(index='native_country',
                               columns='measurement',
                               values='value',
                               aggfunc='mean')
```
