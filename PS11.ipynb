```python
import pandas as pd

df = pd.read_csv("movies_metadata.csv",low_memory=False)
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
df = df[['original_language',
         'vote_average',
         'vote_count',
         'popularity']]
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
df['vote_average'] = pd.to_numeric(df['vote_average'],errors='coerce')
df['vote_count'] = pd.to_numeric(df['vote_count'],errors='coerce')
df['popularity'] = pd.to_numeric(df['popularity'],errors='coerce')
```

```python
df = df.dropna()
```

```python
english_data = df[df['original_language']=='en']

french_data = df[df['original_language']=='fr']
```

```python
print(english_data.head())
```

```python
print(french_data.head())
```

```python
df_concat = pd.concat([english_data,french_data])
```

```python
print(df_concat)
```

```python
df_concat.head()
```

```python
df_sorted = df.sort_values(by='vote_average',ascending=True)
```

```python
df_sorted.head()
```

```python
melted_data = pd.melt(df,
                      id_vars=['original_language'],
                      value_vars=['vote_average',
                                  'vote_count',
                                  'popularity'],
                      var_name='measurement',
                      value_name='value')
```

```python
print(melted_data.head())
```

```python
wide = melted_data.pivot_table(index='original_language',
                               columns='measurement',
                               values='value',
                               aggfunc='mean')
```
