```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Cleavland.csv",header=None)
```

```python
df.columns = ['age',
              'sex',
              'cp',
              'trestbps',
              'chol',
              'fbs',
              'restecg',
              'thalach',
              'exang',
              'oldpeak',
              'slope',
              'ca',
              'thal',
              'num']
```

```python
print(df.head())
```

```python
plt.hist(df['age'])

plt.title("Histogram of Age")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.plot(df['chol'],'.')

plt.title("Dot Plot of Cholesterol")

plt.xlabel("Index")

plt.ylabel("Cholesterol")

plt.show()
```

```python
avg_chol = df.groupby('sex')['chol'].mean()

avg_chol.plot(kind='bar')

plt.title("Average Cholesterol by Sex")

plt.xlabel("Sex")

plt.ylabel("Average Cholesterol")

plt.show()
```

```python
plt.plot(df['thalach'])

plt.title("Line Chart of Maximum Heart Rate")

plt.xlabel("Index")

plt.ylabel("Heart Rate")

plt.show()
```

```python
plt.boxplot(df['chol'])

plt.title("Box Plot of Cholesterol")

plt.show()
```

```python
plt.hist(df['chol'])

plt.scatter(range(len(df['chol'])),
            df['chol'])

plt.title("Histogram and Scatter Plot of Cholesterol")

plt.xlabel("Index")

plt.ylabel("Cholesterol")

plt.show()
```
